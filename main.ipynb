```python

import requests

url = "https://musicbrainz.org/ws/2/artist/"

params = {
    "query": "artist:Radiohead",
    "fmt": "json",
    "limit": 10
}

headers = {
    "User-Agent": "EmergingArtistProject/1.0 (advani.devin@gmail.com)"
}

response = requests.get(
    url,
    params=params,
    headers=headers
)

data = response.json()

print(data)
```

```python

artists = data["artists"]

for artist in artists:
    print(
        artist.get("name"),
        artist.get("country"),
        artist.get("type"),
        artist.get("begin")
    )
```

```python

artist_id = artists[0]["id"]

url = f"https://musicbrainz.org/ws/2/artist/{artist_id}"

params = {
    "fmt": "json",
    "inc": "genres+tags+aliases"
}

response = requests.get(
    url,
    params=params,
    headers=headers
)

artist_details = response.json()

print(artist_details.keys())
```

```python

import pandas as pd

artist_data = []

for artist in artists:

    artist_data.append({
        "name": artist.get("name"),
        "id": artist.get("id"),
        "country": artist.get("country"),
        "type": artist.get("type"),
        "begin": artist.get("begin"),
        "genres": [
            tag["name"] for tag in artist.get("genres", [])
        ]
    })

df = pd.DataFrame(artist_data)

print(df.head())
```
