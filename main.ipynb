````python

import requests

url = "https://musicbrainz.org/ws/2/artist/"

params = {
    "query": "artist:Radiohead",
    "fmt": "json",
    "limit": 10
}

headers = {
    "User-Agent": "EmergingArtistProject/1.0 (your-email@example.com)"
}

response = requests.get(
    url,
    params=params,
    headers=headers
)

data = response.json()

print(data)
````

````python

artists = data["artists"]

for artist in artists:
    print(
        artist.get("name"),
        artist.get("country"),
        artist.get("type"),
        artist.get("begin")
    )
````

````python

artist_id = artists[0]["id"]

url = f"https://musicbrainz.org/ws/2/artist/{artist_id}"

params = {
    "fmt": "json",
    "inc": "genres+tags+aliases"
}

response = requests.get(
    url,
    params=params,
    headers=headers
)

artist_details = response.json()

print(artist_details.keys())
````

````python
import time

url = "https://musicbrainz.org/ws/2/release-group"

params = {
    "artist": artist_id,
    "fmt": "json",
    "limit": 100
}

time.sleep(1.1)

response = requests.get(
    url,
    params=params,
    headers=headers
)

releases = response.json()["release-groups"]

for release in releases:
    print(
        release.get("title"),
        release.get("first-release-date"),
        release.get("primary-type")
    )
````

````python

import pandas as pd

artist_data = []

for artist in artists:

    artist_data.append({
        "name": artist.get("name"),
        "id": artist.get("id"),
        "country": artist.get("country"),
        "type": artist.get("type"),
        "begin": artist.get("begin"),
        "genres": [
            tag["name"] for tag in artist.get("genres", [])
        ]
    })

df = pd.DataFrame(artist_data)

print(df.head())
```

You can then export it:

```python
df.to_csv("musicbrainz_artists.csv", index=False)
```

**Important:** MusicBrainz limits API usage to approximately one request per second. Respect this limit when collecting large datasets. <Cite refs={["turn402142search2"]}/>

# 7. How I'd use this for your emerging artist project

I'd aim to create a dataset with the following structure:

| Feature | Source |
|---|---|
| Artist name | MusicBrainz |
| Country | MusicBrainz |
| Genre | MusicBrainz |
| First release date | MusicBrainz |
| Number of releases | MusicBrainz |
| Record label | MusicBrainz |
| Monthly listeners | Streaming platform data |
| Social media followers | Social media data |
| Listener growth | Historical snapshots |

One important limitation: MusicBrainz is primarily a metadata database. It doesn't provide Spotify monthly listeners, streaming popularity or social media growth.

I'd therefore combine it with other data sources, particularly Last.fm, where available.

# 8. The project I'd work towards

<box border radius="lg" padding={3} gap={3}>
  <box gap={1}>
    <badge color="success">Portfolio project</badge>
    <title size="2xl">Emerging Artist Discovery Engine</title>
    <text color="secondary" size="sm">Music industry analytics and machine learning</text>
  </box>
  <AsyncImage query="dark modern music analytics dashboard artist discovery data visualization" aspectRatio="16:9" width="100%"/>
  **End goal:**

  A searchable database of emerging artists that identifies promising musicians based on:
  - Genre and location.
  - Release frequency.
  - Growth in listeners.
  - Social media engagement.
  - Historical popularity trends.

  **Potential techniques:**
  - Exploratory data analysis
  - K-means clustering
  - Random forest
  - XGBoost
  - Time-series forecasting
</box>

## What I'd do first

I'd actually start with a relatively small experiment:

1. Collect 500 artists from MusicBrainz.
2. Extract their country, genres and release histories.
3. Investigate which genres and countries have the most emerging artists.
4. Combine this with streaming popularity data.
5. Investigate whether release frequency and other features correlate with growth.

**One distinction worth remembering:** MusicBrainz can tell you who an artist is and what they've released. It cannot tell you whether they're about to become successful. That's where combining datasets and building your own models becomes interesting.

Official documentation: <Link url="https://musicbrainz.org/doc/MusicBrainz_API" title="MusicBrainz API documentation"/>.
````
